```python
from pyspark.sql import functions as F
```

```python
silver_df = spark.table("Employee_info.dim_employee")
```

```python

salary_by_department_df = (
    silver_df
    .groupBy("department")
    .agg(
        F.sum("salary").alias("total_salary")
    )
    .orderBy(
        F.col("total_salary").desc()
    )
)

display(salary_by_department_df)
```

```python
employee_count_df = (
    silver_df
    .groupBy("department", "country")
    .agg(
        F.count("employee_id").alias("employee_count")
    )
    .orderBy(
        "department",
        "country"
    )
)

display(employee_count_df)
```

```python
department_country_df = (
    silver_df
    .select(
        "department",
        "country"
    )
    .distinct()
    .orderBy(
        "department",
        "country"
    )
)

display(department_country_df)
```

```python
average_age_df = (
    silver_df
    .groupBy("department")
    .agg(
        F.avg("age").alias("average_age")
    )
    .orderBy("department")
)

display(average_age_df)
```
