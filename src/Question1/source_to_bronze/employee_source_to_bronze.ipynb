```python
employee_input_path = "/Volumes/assignment/bronze/raw_data/Employee-Q1-in.csv"

department_input_path = "/Volumes/assignment/bronze/raw_data/Department-Q1-in.csv"

country_input_path = "/Volumes/assignment/bronze/raw_data/Country-Q1-in.csv"
```

```python
employee_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(employee_input_path)
)
```

```python
display(employee_df)
```

```python
employee_df.printSchema()
```

```python
department_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(department_input_path)
)
department_df.printSchema()
```

```python
country_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(country_input_path)
)

country_df.printSchema()
```

```python
employee_df.write \
    .mode("overwrite") \
    .option("header", True) \
    .csv(
        "/Volumes/assignment/bronze/raw_data/source_to_bronze/employee"
    )
```

```python
department_df.write \
    .mode("overwrite") \
    .option("header", True) \
    .csv(
        "/Volumes/assignment/bronze/raw_data/source_to_bronze/department"
    )
```

```python
country_df.write \
    .mode("overwrite") \
    .option("header", True) \
    .csv(
        "/Volumes/assignment/bronze/raw_data/source_to_bronze/country"
    )
```
